```python
import pandas as pd
import re

def classify_columns(df, cardinality_threshold=10):
    categorical = []
    ordinal     = []
    nominal     = []
    discrete    = []
    continuous  = []

    # Ordinal patterns (deciles, ratings, ranks)
    ordinal_patterns = [
        r"decile$", r"_decile$", r"rating$", r"_rating$",
        r"rank$",   r"_rank$",   r"grade$",  r"tier$",
    ]

    # Continuous patterns (index scores + vehicle counts)
    continuous_patterns = [
        r"^irsd.*score",  r"^irsad.*score", r"^ier.*score", r"^ieo.*score",   # ← only scores
        r"no_vehicles"
    ]

    # Postcode / geo-code → nominal
    geocode_patterns = [
        r"^pcode$", r"postcode$", r"^lat", r"^lon",
        r"postal.area", r"poa.*code", r"lga.*code", r"sa[1-4]",
    ]

    def matches_any(col_lower, patterns):
        return any(re.search(p, col_lower) for p in patterns)

    def is_effectively_numeric(series):
        converted = pd.to_numeric(series, errors="coerce")
        return converted.notna().mean() > 0.5, converted

    for col in df.columns:

        if col.lower() == "objectid":
            continue

        dtype      = df[col].dtype
        n_unique   = df[col].nunique(dropna=True)
        col_lower  = col.lower().strip()
        is_numeric = pd.api.types.is_numeric_dtype(dtype)
        is_text    = dtype == "object" or isinstance(dtype, pd.CategoricalDtype)
        is_bool    = pd.api.types.is_bool_dtype(dtype)

        effective_numeric = False
        converted_series  = None
        if is_text:
            effective_numeric, converted_series = is_effectively_numeric(df[col])

        # ── Priority 1: Continuous patterns ───────────────────────────────────
        if matches_any(col_lower, continuous_patterns):
            continuous.append(col)

        # ── Priority 2: Ordinal patterns ──────────────────────────────────────
        elif matches_any(col_lower, ordinal_patterns):
            ordinal.append(col)

        # ── Priority 3: Geo/postcode → nominal ────────────────────────────────
        elif matches_any(col_lower, geocode_patterns):
            nominal.append(col)

        # ── Priority 4: Object dtype but numeric underneath ────────────────────
        elif is_text and effective_numeric:
            if n_unique <= cardinality_threshold:
                ordinal.append(col)
            else:
                continuous.append(col)

        # ── Priority 5: Cardinality + dtype based ─────────────────────────────
        elif n_unique <= cardinality_threshold:
            if is_numeric:
                ordinal.append(col)
            else:
                categorical.append(col)

        else:
            if pd.api.types.is_float_dtype(dtype):
                continuous.append(col)
            elif pd.api.types.is_integer_dtype(dtype):
                discrete.append(col)
            else:
                categorical.append(col)

    return categorical, ordinal, nominal, discrete, continuous


categorical, ordinal, nominal, discrete, continuous = classify_columns(df5)

print("Categorical:", categorical)
print("Ordinal:    ", ordinal)
print("Nominal:    ", nominal)
print("Discrete:   ", discrete)
print("Continuous: ", continuous)
```

### CATEGORICAL BARCHART

```python
import matplotlib.pyplot as plt
import numpy as np

n = len(categorical)
cols = 3
rows = (n + cols - 1) // cols

fig, axes = plt.subplots(rows, cols, figsize=(10 * cols, 10 * rows))
axes = np.array(axes).flatten()

i = -1
for i, col in enumerate(categorical):
    ax = axes[i]

    value_counts = df5[col].value_counts().head(20)
    colors = plt.cm.Set2(np.linspace(0, 1, len(value_counts)))

    ax.bar(
        value_counts.index.astype(str),
        value_counts.values,
        color=colors
    )

    title = f"{col} (Top 20)" if df5[col].nunique() > 20 else col
    ax.set_title(title, fontsize=12)
    ax.set_xlabel("Category")
    ax.set_ylabel("Count")
    ax.set_xticklabels(value_counts.index.astype(str), rotation=45, ha="right")

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle("Categorical Columns", fontsize=16, fontweight="bold", y=1.01)
plt.tight_layout()
plt.show()
```

```python
cols_to_check = [
    "Usual Resident Population", "IER Score", "IEO Score", "IRSAD Score", "IRSD Score",
    "PCODE", "Longitude", "Latitude", "Tesla_Fast", "IER Decile", "IEO Decile", "IRSAD Decile", "IRSD Decile"
]

threshold = 20  # only show value counts if unique values <= threshold

for col in cols_to_check:
    if col not in df5.columns:
        print(f"⚠️  '{col}' not found in dataframe\n")
        continue
    
    n_unique = df5[col].nunique()
    print(f"{'='*50}")
    print(f"Column: {col} | Unique values: {n_unique}")
    print(f"{'='*50}")
    
    if n_unique <= threshold:
        vc = df5[col].value_counts().reset_index()
        vc.columns = ["Value", "Count"]
        vc["Percentage"] = (vc["Count"] / len(df5) * 100).round(2).astype(str) + "%"
        print(vc.to_string(index=False))
    else:
        print(f"  Skipped — too many unique values ({n_unique})")
    
    print()
```
